```python
# Criando um dicionário de tradução
traducao_colunas = {'ID': 'id',
    'Customer_ID': 'id_cliente',
    'Month': 'mes',
    'Name': 'nome',
    'Age': 'idade',
    'SSN': 'cpf_rg',
    'Occupation': 'ocupacao',
    'Annual_Income': 'renda_anual',
    'Monthly_Inhand_Salary': 'salario_mensal_liquido',
    'Num_Bank_Accounts': 'num_contas_bancarias',
    'Num_Credit_Card': 'num_cartoes_credito',
    'Interest_Rate': 'taxa_juros',
    'Num_of_Loan': 'num_emprestimos',
    'Type_of_Loan': 'tipo_emprestimo',
    'Delay_from_due_date': 'dias_atraso_vencimento',
    'Num_of_Delayed_Payment': 'num_pagamentos_atrasados',
    'Changed_Credit_Limit': 'limite_credito_alterado',
    'Num_Credit_Inquiries': 'num_consultas_credito',
    'Credit_Mix': 'mix_credito',
    'Outstanding_Debt': 'divida_pendente',
    'Credit_Utilization_Ratio': 'taxa_utilizacao_credito',
    'Credit_History_Age': 'idade_historico_credito',
    'Payment_of_Min_Amount': 'pagamento_minimo_fatura',
    'Total_EMI_per_month': 'total_parcelas_mensais',
    'Amount_invested_monthly': 'valor_investido_mensal',
    'Payment_Behaviour': 'comportamento_pagamento',
    'Monthly_Balance': 'saldo_mensal',
    # A coluna exclusiva do arquivo de treino:
    'Credit_Score': 'score_credito' 
}
# Renomeando as colunas
df = df.rename(columns=traducao_colunas)
```

```python
df.columns
```

```python

colunas_limpar = ['idade', 'renda_anual', 'num_pagamentos_atrasados', 'divida_pendente']

for coluna in colunas_limpar:
    # Remove qualquer caractere que não seja número ou ponto
    df[coluna] = df[coluna].astype(str).str.replace(r'[^0-9.]', '', regex=True)
    
    # Converte para numérico (se der erro, vira NaN - Not a Number)
    df[coluna] = pd.to_numeric(df[coluna], errors='coerce')

# Remove linhas que ficaram vazias após a limpeza (opcional, mas ajuda)
df = df.dropna(subset=colunas_limpar)
```

```python
# Ver qual a renda média de quem tem score bom vs ruim, quem ganha mais tem socre melhor?
print(df.groupby('score_credito')['renda_anual'].mean())
```

```python
#Representatividade dos scores
df['score_credito'].value_counts() # good - score bom/ poor - score ruim/ standard - score mediano
```

```python
#Score vs atrasos. Quem atrasa mais tem o score menor?
df.groupby('score_credito')['dias_atraso_vencimento'].mean().round(2)
```

```python
#qual é o mix de crédito mais significativo?
df['mix_credito'].value_counts()
```

```python
# Criando categorias de idade
bins = [0, 25, 45, 65, 120]
labels = ['Jovem (até 25)', 'Adulto (26-45)', 'Sênior (46-65)', 'Idoso (65+)']

df['faixa_etaria'] = pd.cut(df['idade'], bins=bins, labels=labels)

# Quantos "Good", "Standard" e "Poor" existem em cada faixa
analise_idade = df.groupby(['faixa_etaria', 'score_credito']).size().unstack()
print(analise_idade)
```

```python
#renda media anual por idade
analise_idade_renda = df.groupby('idade')['renda_anual'].mean().round(2)
```

```python
print(analise_idade_renda)
```

```python
# Filtrando apenas idades que fazem sentido. Entre 18 e 100 anos
df = df[(df['idade'] >= 18) & (df['idade'] <= 100)]


analise_idade_renda = df.groupby('idade')['renda_anual'].mean()
print(analise_idade_renda)
```

```python
#nTaxa Média de Utilização de % de limite de Crédito 
print(df.groupby('score_credito')['taxa_utilizacao_credito'].mean().round(2))
```

```python
# Cruzamento de Comportamento vs Score
matriz_comportamento = pd.crosstab(df['comportamento_pagamento'], df['score_credito'])
print("\nMatriz de Comportamento de Pagamento:")
print(matriz_comportamento)
```

```python
# Limpeza da coluna num_emprestimos
# Remove qualquer coisa que não seja número e converte para numérico
df['num_emprestimos'] = pd.to_numeric(df['num_emprestimos'].astype(str).str.replace(r'[^0-9]', '', regex=True), errors='coerce')
```

```python
# Média de empréstimos por score
print(df.groupby('score_credito')['num_emprestimos'].mean().round(2))
```

```python
# Lista das colunas que precisam ser números para o resumo
colunas_numericas = [
    'idade', 'renda_anual', 'divida_pendente', 
    'taxa_utilizacao_credito', 'num_emprestimos', 'valor_investido_mensal'
]

for col in colunas_numericas:
    # Remove tudo que não for número ou ponto, depois converte
    df[col] = pd.to_numeric(df[col].astype(str).str.replace(r'[^0-9.]', '', regex=True), errors='coerce')

# Filtra as idades novamente por segurança
df = df[(df['idade'] >= 18) & (df['idade'] <= 100)]
```

```python
# Criando o DataFrame de resumo comparativo
resumo_perfil = df.groupby('score_credito').agg({
    'idade': 'mean',
    'renda_anual': 'mean',
    'divida_pendente': 'mean',
    'taxa_utilizacao_credito': 'mean',
    'num_emprestimos': 'mean',
    'valor_investido_mensal': 'mean'
}).round(2)

# Transpondo a tabela para facilitar a leitura (Score vira as colunas)
resumo_perfil = resumo_perfil.T

print("=== RESUMO ESTRATÉGICO DE PERFIL DE CRÉDITO ===")
print(resumo_perfil)
```
